```python
import pandas as pd
import numpy as np
import plotly.express as px

# ------------------------------------------
# 0. PREPARE DATA
# ------------------------------------------

df["acquisition_date"] = pd.to_datetime(df["acquisition_date"])
df["cancellation_month"] = pd.to_datetime(df["cancellation_month"], errors="coerce")

df["acquisition_month"] = df["acquisition_date"].dt.to_period("M").dt.to_timestamp()

# ------------------------------------------
# 1. COHORT SIZE (NEW USERS PER MONTH)
# ------------------------------------------

cohort_size = (
    df.groupby("acquisition_month")
      .agg(new_users=("user_id", "count"))
      .reset_index()
)

# ------------------------------------------
# 2. MONTHS SINCE ACQUISITION
# ------------------------------------------

df["month_churned"] = (
    (df["cancellation_month"].dt.year - df["acquisition_month"].dt.year) * 12 +
    (df["cancellation_month"].dt.month - df["acquisition_month"].dt.month)
)

df.loc[df["month_churned"] < 0, "month_churned"] = np.nan

# ------------------------------------------
# 3. CHURN COUNTS
# ------------------------------------------

cohort_data = (
    df.dropna(subset=["month_churned"])
      .groupby(["acquisition_month", "month_churned"])
      .agg(users=("user_id", "count"))
      .reset_index()
)

cohort_data["month_churned"] = cohort_data["month_churned"].astype(int)

# ------------------------------------------
# 4. FULL GRID (0–12 MONTHS)
# ------------------------------------------

max_month = 12
tenure_range = np.arange(0, max_month + 1)

full_index = pd.MultiIndex.from_product(
    [
        cohort_data["acquisition_month"].unique(),
        tenure_range
    ],
    names=["acquisition_month", "month_churned"]
)

cohort_data = (
    cohort_data
    .set_index(["acquisition_month", "month_churned"])
    .reindex(full_index, fill_value=0)
    .reset_index()
)

# ------------------------------------------
# 5. MERGE COHORT SIZE
# ------------------------------------------

cohort_data = cohort_data.merge(cohort_size, on="acquisition_month", how="left")

# ------------------------------------------
# 6. CUMULATIVE CHURN + RETENTION
# ------------------------------------------

cohort_data = cohort_data.sort_values(["acquisition_month", "month_churned"])

cohort_data["cumulative_churn"] = (
    cohort_data.groupby("acquisition_month")["users"].cumsum()
)

cohort_data["churn_rate"] = cohort_data["cumulative_churn"] / cohort_data["new_users"]
cohort_data["retention_rate"] = 1 - cohort_data["churn_rate"]

# ------------------------------------------
# 7. PIVOT TABLE (HEATMAP FORMAT)
# ------------------------------------------

heatmap_data = cohort_data.pivot(
    index="acquisition_month",
    columns="month_churned",
    values="retention_rate"
)

heatmap_data = heatmap_data.sort_index()

# ------------------------------------------
# 8. HEATMAP VISUALIZATION
# ------------------------------------------

fig = px.imshow(
    heatmap_data,
    text_auto=".2f",
    aspect="auto",
    color_continuous_scale="Blues",
    title="Cohort Retention Heatmap (12 Months)"
)

fig.update_layout(
    xaxis_title="Months Since Acquisition",
    yaxis_title="Acquisition Month"
)

fig.show()
```
