## Part 2

```python
sales = [120, 250, 80, 310, 95]

# total
total = sum(sales)

# average
avg = total / len(sales)

# max without max()
max_value = sales[0]
for s in sales:
    if s > max_value:
        max_value = s

# count >150
count = len([s for s in sales if s > 150])

# +15%
new_sales = [s * 1.15 for s in sales]

total, avg, max_value, count, new_sales
```
