```python
import pandas as pd
```

```python
df_products = pd.read_csv('../data/raw/products.csv')
print(df_products.columns)
```

```python
df_deparments = pd.read_csv("../data/processed/departments_t.csv")
print(df_deparments.columns)
```

```python
df_products_departments = pd.merge(
    df_products,
    df_deparments,
    on= 'department_id',
    how= 'left'
    )
df_products_departments.head(7)
```

```python
df_aisles = pd.read_csv('../data/raw/aisles.csv')
```

```python
df_products_enriched= pd.merge(
    df_products_departments,
    df_aisles,
    on= 'aisle_id',
    how= 'left'
    )
df_products_enriched.head()
```

```python
df_orders_train= pd.read_csv('../data/raw/order_products_train.csv')
```

```python
df_order_products_train= pd.read_csv('../data/raw/order_products_train.csv')
```

```python
df_train_transactions= pd.merge(
    df_orders_train,
    df_order_products_train,
    on= 'order_id',
    how= 'inner'
)
df_train_transactions.head() 
```

```python
df_train_transactions.rename(columns={'product_id_y':'product_id'}, inplace=True)
```

```python
df_train_transactions.drop(columns=['product_id_x', 'add_to_cart_order_x', 'reordered_x'], inplace=True)
```

```python
df_train_analytic= pd.merge(
    df_train_transactions,
    df_products_enriched,
    on='product_id',
    how="left"
)
df_train_analytic.head()
```
